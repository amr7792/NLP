```python
data['clean_data']=data['text'].apply(lambda x: re.sub(r'[^A-Za-z]+',' ',x))
```

```python
data['clean_data']=data['clean_data'].apply(lambda x: x.lower())
```

```python
data['clean_data']=data['clean_data'].apply(lambda x: x.strip())
```

```python
print("Actual data=======> {}".format(data['text'][1]))
```

```python
print("Cleaned data=======> {}".format(data['clean_data'][1]))
```

```python
from nltk.corpus import stopwords
stopwords=set(stopwords.words('english'))
```

```python
data['clean_data']=data['clean_data'].apply(lambda x: ' '.join([words for words in x.split() if words not in stopwords]))
```

```python
data['clean_data'][6]
```

```python
data['labels']=data.apply(lambda col: [col['gender'],str(col['age']),col['topic'],col['sign']], axis=1)
```

```python
data.head()
```

```python
data=data[['clean_data','labels']]
```

```python
data
```

```python
X=data['clean_data']
```

```python
Y=data['labels']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorizer=CountVectorizer(binary=True, ngram_range=(1,2))
```

```python
X=vectorizer.fit_transform(X)
```

```python
X[1]
```

```python
vectorizer.get_feature_names_out()[:5]
```

```python
label_counts=dict()

for labels in data.labels.values:
    for label in labels:
        if label in label_counts:
            label_counts[label]+=1
        else:
            label_counts[label]=1
```

```python
label_counts
```

```python
from sklearn.preprocessing import MultiLabelBinarizer
binarizer=MultiLabelBinarizer(classes=sorted(label_counts.keys()))
```

```python
binarizer
```

```python
Y=binarizer.fit_transform(data.labels)
```

```python
from sklearn.model_selection import train_test_split
```

```python
Xtrain,Xtest,Ytrain,Ytest=train_test_split(X,Y,test_size=0.2)
```

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression(solver='lbfgs')
```

```python
model=OneVsRestClassifier(model)
```

```python
model.fit(Xtrain,Ytrain)
```

```python
Ypred=model.predict(Xtest)
```

```python
Ypred_inversed = binarizer.inverse_transform(Ypred)
y_test_inversed = binarizer.inverse_transform(Ytest)
```

```python
for i in range(5):
    print('Text:\t{}\nTrue labels:\t{}\nPredicted labels:\t{}\n\n'.format(
        Xtest[i],
        ','.join(y_test_inversed[i]),
        ','.join(Ypred_inversed[i])
    ))
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
from sklearn.metrics import average_precision_score
from sklearn.metrics import recall_score

def print_evaluation_scores(Ytest, Ypred):
    print('Accuracy score: ', accuracy_score(Ytest, Ypred))
    print('F1 score: ', f1_score(Ytest, Ypred, average='micro'))
    print('Average precision score: ', average_precision_score(Ytest, Ypred, average='micro'))
    print('Average recall score: ', recall_score(Ytest, Ypred, average='micro'))
```

```python
print_evaluation_scores(Ytest, Ypred)
```

*   **Low Accuracy (31.7%):** The model correctly classifies only about 32% of instances, indicating poor overall performance.
*   **Moderate F1-score (64%):** Suggests a reasonable balance between precision and recall, but still leaves room for improvement.
*   **Low Average Precision (46%):** Indicates a moderate number of false positives.
*   **Moderate Recall (53%):** The model captures slightly over half of the true positive instances.
*   **High False Positives/Negatives:** The combination of low accuracy and moderate F1 suggests a substantial number of both false positives and false negatives, with a tendency towards more false positives.
*   **Potential Data Imbalance:** The difference between precision and recall *could* indicate imbalanced data.
*   **Model Improvement Needed:** Feature engineering, trying different algorithms, hyperparameter tuning, or addressing data imbalance should be explored.
